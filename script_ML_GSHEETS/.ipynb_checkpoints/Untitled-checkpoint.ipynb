```python
# Coluna de Conversão dos índices
def coluna_para_indice(coluna):
    coluna = coluna.upper()
    indice = 0
    for letra in coluna:
        indice = indice * 26 + (ord(letra) - ord('A') + 1)
    return indice
```

```python
# Nome coluna
coluna = coluna_para_indice(input())
```
