```python
# Instalando o pacote do Kaggle
# !pip install kagglehub

# Download do dataset
import kagglehub

# Download latest version
path = kagglehub.dataset_download("rockyt07/social-media-user-analysis")

print("Path to dataset files:", path)
```

```python
# Bibliotecas
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(f'{path}/instagram_usage_lifestyle.csv')
```

```python
# Amostragem
df.head()
```

```python
df.columns
```

<h3> 2️⃣ Análise Exploratória de Acordo com o Tipo de Variável

```python
df.shape
```

```python
# Visualizar se há alguma coluna com dados faltantes
df.info()
```

```python
# Variáveis Quantitativas x Qualitativas
df.dtypes.value_counts()
```

```python
# Separando as features de acordo com o tipo
var_qualitativas = df.select_dtypes(include = ['object'])
var_quantitativas = df.select_dtypes(include = ['float64','int64'])
```

```python
# Removendo a coluna app_name e user_id
var_qualitativas.drop(columns='app_name', inplace = True)
var_quantitativas.drop(columns='user_id', inplace = True)
```

## Explorando dados Qualitativos

```python
var_qualitativas.head()
```

```python
# Plotando urban_rural
fig, axes = plt.subplots(1,1, figsize = (10,4))
fig.suptitle('Distribuição de urban_rural')
sns.countplot(data = var_qualitativas, x = 'urban_rural')
```

```python
# Criando tabela de frequencia de urban_rural
tab_freq_urban_rural = var_qualitativas['urban_rural'].value_counts()
tab_freq_urban_rural
```

```python
# Plotando frequencia de genero
fig.suptitle('Distribuição de Gênero')
sns.countplot(data = var_qualitativas, x = 'gender')
```

```python
# Criando tabela de frequencia de genero
tab_freq_gender = var_qualitativas['gender'].value_counts()
tab_freq_gender
```

```python
# Plotando a coluna employment_status
fig, axes = plt.subplots(1,1, figsize = (10,5)) 
fig.suptitle('Distribuição de Status de Emrpegados')
sns.countplot(data = var_qualitativas, x = 'employment_status')
```

```python
# Criando tabela de Frequencia de Status empregados
tab_freq_employment_status = var_qualitativas['employment_status'].value_counts()
tab_freq_employment_status
```

```python
# Plotando a coluna education_level
fig, axes = plt.subplots(1,1, figsize = (10,5)) 
fig.suptitle('Distribuição de Status de Emrpegados')
sns.countplot(data = var_qualitativas, x = 'education_level')
```

```python
# Criando tabela de Frequencia de education_level
tab_freq_education_level = var_qualitativas['education_level'].value_counts()
tab_freq_education_level
```

```python
# Plotando a variavel relationship_status
fig.suptitle('Distribuição de relationship_status')
sns.countplot(data = var_qualitativas, x = 'relationship_status')
```

```python
# Criando tabela de Frequencia de education_level
tab_freq_relationship_status = var_qualitativas['relationship_status'].value_counts()
tab_freq_relationship_status
```

```python
# Verificando a correlação de gênero "Prefer not to say"
df.groupby('gender')['user_engagement_score'].mean().sort_values()
```

```python
df.boxplot(column = 'user_engagement_score', by = 'gender')
plt.title('Target por Gênero')
plt.suptitle('')
plt.show()
```

```python
# Verificando a correlação de relationship_status "widowed"
df.groupby('relationship_status')['user_engagement_score'].mean().sort_values()
```

```python
df.boxplot(column = 'user_engagement_score', by = 'relationship_status')
plt.title('Target por relationship_status')
plt.suptitle('')
plt.show()
```

> Apesar de representarem boas médias em relação a variável alvo, por representarem uma amostragem muito pequena,

> Vamos remover os valores 'Prefer not to say' e 'Widowed'

```python
# Removendo categorias com baixo número para evitar ruídos no modelo
df = df[df['gender'] != 'Prefer not to say']
df = df[df['relationship_status'] != 'widowed']
```
